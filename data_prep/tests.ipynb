```python
import pandas as pd 
import numpy as np
```

```python
climate_df = pd.read_csv("../raw_data/climate.csv.gz")
climate_updated = pd.read_csv("../raw_data/climate_update_2026.csv.gz")
processed_climate = pd.read_csv("../processed_data/climate/climate.csv.gz")
forecasting_climate = pd.read_csv("../raw_data/forecasting_climate.csv.gz")
forecastingg_climate_updated = pd.read_csv("../raw_data/forecasting_climate_update_2026.csv.gz")
ocean_climate = pd.read_csv("../raw_data/ocean_climate_oscillations.csv.gz")
ocean_climate_updated = pd.read_csv("../raw_data/ocean_climate_oscillations_update_2026.csv.gz")
```

```python
ocean_climate_updated.head(2)
```

```python
climate_df['date'] = pd.to_datetime(climate_df['date'])
climate_updated['date'] = pd.to_datetime(climate_updated['date'])
forecasting_climate['reference_month'] = pd.to_datetime(forecasting_climate['reference_month'], format='mixed')
forecastingg_climate_updated['reference_month'] = pd.to_datetime(forecastingg_climate_updated['reference_month'], format='mixed')
# Find the dates
oldest_date = climate_df['date'].min()
newest_date = climate_df['date'].max()
print(f"Oldest date: {oldest_date}, Newest date: {newest_date}")

oldest_date_updated = climate_updated['date'].min()
newest_date_updated = climate_updated['date'].max()
print(f"Oldest date in updated: {oldest_date_updated}, Newest date in updated: {newest_date_updated}")

oldest_date_forecasting = forecasting_climate['reference_month'].min()
newest_date_forecasting = forecasting_climate['reference_month'].max()
print(f"Oldest date in forecasting: {oldest_date_forecasting}, Newest date in forecasting: {newest_date_forecasting}")

oldest_date_forecasting_updated = forecastingg_climate_updated['reference_month'].min()
newest_date_forecasting_updated = forecastingg_climate_updated['reference_month'].max()
print(f"Oldest date in forecasting updated: {oldest_date_forecasting_updated}, Newest date in forecasting updated: {newest_date_forecasting_updated}")

oldest_date_ocean = ocean_climate['date'].min()
newest_date_ocean = ocean_climate['date'].max()
print(f"Oldest date in ocean climate: {oldest_date_ocean}, Newest date in ocean climate: {newest_date_ocean}")

oldest_date_ocean_updated = ocean_climate_updated['date'].min()
newest_date_ocean_updated = ocean_climate_updated['date'].max()
print(f"Oldest date in ocean climate updated: {oldest_date_ocean_updated}, Newest date in ocean climate updated: {newest_date_ocean_updated}")
```
